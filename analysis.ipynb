# 삼성전자(005930.KS) 시계열 데이터 분석 및 인사이트 도출

본 노트북은 2024년부터 최근까지의 삼성전자 일별 주가/거래량 데이터를 기반으로 시계열 분석(이동평균, 변동성, 수익률, 성분 분해, 베이스라인 예측)을 수행합니다.

```python
import os
import sys
sys.path.append('src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = 'Malgun Gothic' if os.name == 'nt' else 'AppleGothic'
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style="whitegrid", font=plt.rcParams['font.family'])

from analysis_utils import (
    load_and_preprocess_data,
    calculate_technical_indicators,
    decompose_time_series,
    simple_baseline_forecast
)
```

## 1. 데이터 로드 및 기술적 지표 계산

```python
df_raw = load_and_preprocess_data('data/samsung_stock_2024_present.csv')
df = calculate_technical_indicators(df_raw)
print(f"총 데이터 수: {len(df)}개")
df.head()
```

## 2. 주가 추이 및 이동평균선 / 볼린저 밴드 시각화

```python
fig, ax = plt.subplots(figsize=(14, 7))
ax.plot(df['Date'], df['Close'], label='종가 (Close)', color='#1f77b4', linewidth=1.8)
ax.plot(df['Date'], df['SMA_20'], label='20일 이동평균 (SMA 20)', color='#ff7f0e', linestyle='--')
ax.plot(df['Date'], df['SMA_50'], label='50일 이동평균 (SMA 50)', color='#2ca02c', linestyle='-.')
ax.fill_between(df['Date'], df['Lower_Band'], df['Upper_Band'], color='#1f77b4', alpha=0.12, label='볼린저 밴드 (±2 Std)')
ax.set_title("삼성전자 주가 추이 및 이동평균 / 볼린저 밴드 (2024~현재)", fontsize=15, fontweight='bold')
ax.legend()
plt.show()
```

## 3. 시계열 성분 분해 (Seasonal Decomposition)

```python
decomp_df, decomposition = decompose_time_series(df, period=20)
fig = decomposition.plot()
fig.set_size_inches(14, 9)
plt.show()
```

## 4. 향후 30 영업일 베이스라인 예측

```python
forecast_df = simple_baseline_forecast(df, forecast_days=30)
forecast_df.head(10)
```
